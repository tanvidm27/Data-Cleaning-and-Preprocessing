# Cleaning Data

### Check Datatype

```python
Sales.dtypes
```

```python
Sales.columns
```

- Drop Unwanted Columns

```python
Sales.drop(columns=["Inventory_Level","Promotion_Code"], inplace=True)
```

```python
Sales.drop(columns=["Sales_Representative","Return_Reason"], inplace=True)
```

```python
Sales.drop(columns=["Return_Flag","Delivery_Days"], inplace=True)
```

```python
Sales.drop(columns=["Order_Year","Product_Subcategory"], inplace=True)
```

```python
Sales.drop(columns=["Shipping_Method","Region"], inplace=True)
```

```python
Sales.drop(columns=["Order_Time","Customer_Rating"], inplace=True)
```

```python
Sales.columns
```

- Change Data type

```python
Sales["Order_Date"]=pd.to_datetime(Sales["Order_Date"],errors="coerce")
```

```python
Sales["Customer_Age"]=Sales["Customer_Age"].astype("Int64")
```

```python
Sales.dtypes
```

- Check duplicate and remove

```python
Sales.duplicated().sum()
```

```python
Sales.drop_duplicates()
```

- Fill Null Values

```python
Sales.isnull().sum()
```

```python
Sales["Customer_Age"]=Sales["Customer_Age"].fillna(Sales["Customer_Age"].median())
```

```python
Sales["Payment_Method"]=Sales["Payment_Method"].fillna("Unknown")
```

```python
Sales["Customer_Gender"]=Sales["Customer_Gender"].fillna(Sales["Customer_Gender"].mode())
```

```python
Sales["Customer_Gender"]=Sales["Customer_Gender"].fillna("Unknown")
```

```python
Sales["Product_Name"]=Sales["Product_Name"].fillna("Unknown Customer")
Sales["Customer_Name"]=Sales["Customer_Name"].fillna("Unknown Customer")
```
